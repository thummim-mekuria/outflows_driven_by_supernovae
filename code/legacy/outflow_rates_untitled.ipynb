```python
def make_mvpetot_f_ahwc(dat):
    fixer = 1#np.array([1e-1,1e-1,1e-1,1e-1,1,1])[:,np.newaxis,np.newaxis] #theres a extra factor of 10 in the N256 runs
    ms = dat[:,:,:,0]*fixer
    vs = dat[:,:,:,1]/1000 #in terms of 1000km/s
    ps = dat[:,:,:,2]*fixer
    ek = dat[:,:,:,3]*(u.solMass*u.km**2/u.s**2).to(u.erg)*fixer
    et = dat[:,:,:,4]*(u.solMass*u.km**2/u.s**2).to(u.erg)*fixer
    
    t20_vals_mpe = np.array([np.cumsum(arr[:,80:,:],axis=1)[:,-1,:] for arr in [ms,ps,ek+et]])
    t20_v = np.nanmedian(vs[:,80:,:],axis=1)
    
    final_vals = np.concatenate((t20_vals_mpe[0][np.newaxis,:],
                                 t20_v[np.newaxis,:],
                                 t20_vals_mpe[1:])).swapaxes(1,2)
    
    mvpetot_f_ahwc = np.stack((final_vals[:,0,:], 
                               final_vals[:,1,:],
                               (final_vals[:,2,:]+final_vals[:,3,:]),
                               final_vals[:,4,:])).swapaxes(0,1)

    # fixed = mvpetot_f_ahwc*np.array([1,1,1e-1,1e-1,1,1])[np.newaxis,np.newaxis,:] 
    return mvpetot_f_ahwc
```
